```python
def audit_data(df):
    report= pd.DataFrame({
        'Type': df.dtypes,
        'Valeurs Nulles': df.isnull().sum(),
        '% Nulles': (df.isnull().sum() / len(df)) * 100,
        'Valeurs Uniques': df.nunique()
    })
    problems = report[(report['Valeurs Nulles'] > 0) | (report['Type'] == 'object')]
    print(problems)
    return report
report=audit_data(df)
print(report)
```

# Suppression des colonnes redondantes générées par le merge

```python
# On identifie toutes les colonnes qui contiennent le mot 'stats' ou 'Born'
cols_to_drop = [c for c in df.columns if 'stats' in c or 'Born' in c]

# On les supprime d'un coup
df_clean = df.drop(columns=cols_to_drop)

# On vérifie le résultat
print(f"Colonnes restantes : {len(df_clean.columns)}")
print(df_clean.columns.tolist())
```

# Exclusion des gardiens de but

```python
df_clean=df_clean[df_clean['Pos']!='GK']
```

# Supprime les colonnes où TOUTES les valeurs sont nulles (NaN)

```python

df_clean = df_clean.dropna(axis=1, how='all')

# Vérification immédiate
print(f"Nombre de colonnes après nettoyage automatique : {len(df_clean.columns)}")
```

# Suppression des colonnes spécifiques non pertinentes

```python
cols_to_drop=['Rk','OG','unSub',"Mn/Start"]

df_clean.drop(columns=[c for c in cols_to_drop if c in df_clean.columns], inplace=True)

print(f"Nettoyage terminé. Il reste {len(df_clean.columns)} colonnes.")


```

# Conserver uniquement les joueurs fiables (> 900 minutes)

```python
df_clean=df_clean[df_clean['Min']>=900]
```

```python
df_clean.isnull().sum()
```

# Remplacer les NaN par 0 uniquement pour les variables numériques

```python
# 1. On sélectionne uniquement les colonnes qui contiennent des nombres (int ou float)
numeric_cols = df_clean.select_dtypes(include=['float64', 'int64']).columns

# 2. On remplace les NaN par 0 uniquement dans ces colonnes
df_clean[numeric_cols] = df_clean[numeric_cols].fillna(0)

# 3. Vérification : ça doit afficher 0
print(f"Nombre de valeurs nulles restantes : {df_clean.isnull().sum().sum()}")
```

#  FEATURE ENGINEERING (Création des stats Per 90)

```python

cols_to_normalize = ['Gls', 'Ast', 'Sh', 'SoT', 'Int', 'TklW', 'Crs']

for col in cols_to_normalize:

    new_col_name = f"{col}_90"

    df_clean[new_col_name] = (df_clean[col] / df_clean['Min']) * 90


print(df_clean[['Player', 'Gls', 'Gls_90']].head())
```

```python
output_file = '../data/football_data_cleaned.csv'
df_clean.to_csv(output_file, index=False)
```
